# ML Challenge CS_CLUB — reproducible solution
This notebook runs the repository's verified pipeline. It uses only the organizer-provided CSV files, optimizes Macro F1 with out-of-fold predictions, and creates `submission.csv`.

```python
from pathlib import Path
import platform, subprocess, sys

candidates = [
    Path('/kaggle/input/ml-challenge-cs-club'),
    Path('data'),
    Path(r'C:/Users/saite/Downloads/ml-challenge-cs-club'),
]
DATA_DIR = next((p for p in candidates if (p / 'train.csv').exists()), None)
if DATA_DIR is None:
    raise FileNotFoundError('Set DATA_DIR to the folder containing train.csv, test.csv, and sample_submission.csv')
print('Python:', platform.python_version())
print('Data:', DATA_DIR.resolve())
```

```python
command = [sys.executable, 'solution.py', '--data-dir', str(DATA_DIR), '--output-dir', 'artifacts']
subprocess.run(command, check=True)
```

```python
import json
import pandas as pd

summary = json.loads(Path('artifacts/cv_summary.json').read_text())
display(pd.DataFrame(summary['candidate_results']).sort_values('selection_score', ascending=False))
print('Selected:', summary['selected_model'])
print('SHA-256:', summary['submission_sha256'])
submission = pd.read_csv('submission.csv')
display(submission.head())
display(submission.tail())
display(submission['target_popularity'].value_counts().sort_index())
```
